# Conditionals

- Every condition can be expressed as a Boolean operator (True or False)
- Logical operators are:
  - and
  - or

```python
name = ""
print(name)
print(bool(name)) #When false, means data is blank
```

## If Statement

```python
x = 10

if x > 5:
  print("x is greater than 5")
```

### Else Statement

```python
x = 3

if x > 5:
  print("x is greater than 5")
else:
  print("x is not greater than 5")
```

```python
x = 81

if x > 90:
  print("A")
if x > 80:
  print("B")
if x > 70:
  print("C")
else:
  print("Summer School") #Fufills both conditions ind. so prints both
```

```python
x = 81

if x > 90:
  print("A")
elif x > 80: # Makes the condition exclusive so it only prints one
  print("B")
elif x > 70:
  print("C")
else:
  print("Summer School")
```

# Logical Operations

```python
age = 25
income = 50000

if age > 18 and income > 40000: # Makes sure appliee is an adult and makes over 40k
  print("Approved")
```

```python
age = 25
income = 5000000

if (age > 18 and income > 40000) or income > 4000000: # Makes sure appliee is an adult or going to pay
  print("Approved")
```

```python
hours_worked = 40
perf_score = 85
tenure = 2

if (hours_worked >= 40 and perf_score >80) or tenure >= 2:
  print("Promoted!")
else:
  print("Not")
```

## Nested Conditions

```python
x = 3
y = 4

if x==y:
  print("x and y are the same")
else:
  if x < y:
    print("x is smaller")
  else:
    print("y is smaller")
```
